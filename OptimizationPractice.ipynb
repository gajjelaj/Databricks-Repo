```python
customers_data = [
    (1, "Ravi", 101),
    (2, "Priya", 102),
    (3, "John", 101),
    (4, "Anil", 103),
    (5, "Sara", 102),
    (6, "Mike", 101),
    (7, "David", 103),
    (8, "Neha", 102),
]

customers = spark.createDataFrame(
    customers_data, ["customer_id", "customer_name", "dept_id"]
)
```

```python
departments_data = [(101, "IT"), (102, "HR"), (103, "Finance")]

departments = spark.createDataFrame(departments_data, ["dept_id", "dept_name"])
```

```python
result = customers.join(departments, on="dept_id", how="inner")

result.show()
```

```python
from pyspark.sql.functions import broadcast

result = customers.join(broadcast(departments), on="dept_id", how="inner")

result.show()
```

```python
data = [
    ("A", "ADF", 80),
    ("A", "DB", 70),
    ("B", "ADF", 60),
    ("B", "DB", 50),
    ("C", "ADF", 40),
    ("C", "DB", 20),
]

columns = ["student", "subject", "marks"]

df = spark.createDataFrame(data, columns)
```

```python
df.display()
```

```python
from pyspark.sql.functions import *
```

```python
df1 = (
    df.groupBy("student")
    .agg(avg("marks").alias("avg_marks"))
    .withColumn(
        "Result",
        when(col("avg_marks") > 70, "Distn")
        .when(col("avg_marks") > 60, "pass")
        .when(col("avg_marks") < 60, "Fail"),
    )
)
df1.display()
```

```python
data = [
    (1, "John", "IT", 60000),
    (2, "Alice", "HR", 50000),
    (3, "Bob", "IT", 75000),
    (4, "David", "Finance", 65000),
    (5, "Emma", "HR", 55000),
    (6, "Mike", "Finance", 80000),
]

columns = ["EmployeeID", "Name", "Department", "Salary"]

df = spark.createDataFrame(data, columns)
```

```python
df.show()
```

```python
df.printSchema()
```

```python
df1 = df.select("Name", "Salary")
df1.show()
```

```python
df.filter(col("Salary") > 60000).show()
```

```python
df.filter((col("Salary") > 60000) & (col("Department") == "IT")).show()
```

```python
df1 = (
    df.withColumn("Bonus", col("Salary") * 0.10).withColumn(
        "Total_Sal", col("Salary") + col("Bonus")
    )
).show()
```

```python
df.select(sum(col("Salary")).alias("Total_Sal")).show()
```

```python
df.groupBy(col("Department")).agg(sum(col("Salary"))).alias("Total_Dep_Sal").show()
```

```python
df.show()
```

```python
df.groupBy("Department").agg(
    count("*").alias("Emp_Count"),
    sum("Salary").alias("Total_Sal"),
    avg("Salary").alias("avg_Sal"),
    max("Salary").alias("Max_Sal"),
).orderBy(col("Total_Sal").desc()).show()
```

```python
df.filter((col("Salary") > 55000) & (col("Department") == "IT")).show()
```

```python
df.filter((col("Department") == "IT") | (col("Department") == "Finance")).show()
```

```python
# Adding bonus column
df.withColumn("Bonus", col("Salary") * 0.10).show()
```

```python
df.withColumn(
    "SalaryLevel",
    when(col("Salary") >= 70000, "High")
    .when(col("Salary") >= 55000, "Medium")
    .when(col("Salary") < 55000, "Low")
    .otherwise("NA"),
).show()
```

```python
df.groupBy("Department").agg(sum("Salary").alias("Dept_sal")).show()
```

```python
df.groupBy("Department").agg(
    count("*").alias("Dept_count"),
    sum("Salary").alias("Dept_Total_Sal"),
    avg("Salary").alias("Dept_avg_sal"),
    max("Salary").alias("Dept_max_sal"),
).show()
```

```python
df.groupBy("Department").agg(avg("Salary").alias("avg_sal")).filter(
    col("avg_sal") > 60000
).show()
```

```python
df.groupBy("Department").agg(sum("Salary").alias("Total_Sal")).orderBy(
    col("Total_sal").desc()
).show()
```
