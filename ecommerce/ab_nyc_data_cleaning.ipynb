### READ DATA FROM ADLS LANDING CONTAINER

```python
df = spark.read.format("csv")\
    .option("header", "true")\
    .option("inferSchema", "true")\
    .load("abfss://landing@stprojectjaydev01.dfs.core.windows.net/ab_nyc/AB_NYC_2019.csv")
```

```python
df.show()
```

```python
df.count()
```

```python
#Check Schema
df.printSchema()
```

### Check if there are any Nulls in the columns

```python
from pyspark.sql.functions import col, sum, when

df.select([sum(col(c).isNull().cast("int")).alias(c) for c in df.columns]).show()
```

```python
df.dropna().count()
```

```python
df=df.dropna()
```

```python
df.select([sum(col(c).isNull().cast("int")).alias(c) for c in df.columns]).show()
```

```python
df.filter(col("name").isNull()).show()
```

### Check if there any duplicates in the data

```python
df.groupBy(col("id")).count().filter(col("count")>1).show()
```

### Drop Duplicates

```python
df1=df.dropDuplicates(["id"])
df1.count()
```

```python
from pyspark.sql.functions import col

df1.filter(~col("id").rlike("^[0-9]+$")).show()
```

```python
df2=df1.filter(col("id").rlike("^[0-9]+$"))
```

```python
df2.count()
```
